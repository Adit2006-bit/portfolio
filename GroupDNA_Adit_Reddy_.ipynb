```python
import numpy as np
from datetime import datetime
import string
FILE_PATH = "/content/hostel_bois.txt"
with open(FILE_PATH, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
#Feature 1
messages = []
system_count = 0
media_count = 0
deleted_count = 0
for line in lines:
    line = line.strip()
    if line == "":
        continue
    if " - " not in line:
        continue
    try:
        timestamp, rest = line.split(" - ", 1)
        if ": " not in rest:
            system_count += 1
            continue
        sender, message = rest.split(": ", 1)
        media = False
        deleted = False
        if message == "<Media omitted>":
            media = True
            media_count += 1
        if message == "This message was deleted":
            deleted = True
            deleted_count += 1
        messages.append({
            "timestamp": timestamp,
            "sender": sender,
            "message": message,
            "media": media,
            "deleted": deleted
        })
    except:
        system_count += 1

print("Parsed Messages:", len(messages))
print("System Messages:", system_count)
print("Media Messages:", media_count)
print("Deleted Messages:", deleted_count)
participants = set()
for msg in messages:
    participants.add(msg["sender"])

print(participants)
print("Total Participants:", len(participants))
message_count = {}
for msg in messages:
    sender = msg["sender"]
    if sender not in message_count:
        message_count[sender] = 0
    message_count[sender] += 1
sorted_people = sorted(
    message_count.items(),
    key=lambda x: x[1],
    reverse=True
)
for person, count in sorted_people:
    print(person, count)
#Feature 2
    print("="*60)
print("GROUP OVERVIEW")
print("="*60)
print("Participants :", len(participants))
print("Messages :", len(messages))
print()
print("Messages Per Person")
for person, count in sorted_people:
    percent = count/len(messages)*100
    print(f"{person:10} {count:5} ({percent:.1f}%)")
#Feature 3
day_count = {}
for msg in messages:
    date = msg["timestamp"].split(",")[0]
    if date not in day_count:
        day_count[date] = 0
    day_count[date] += 1
# Find busiest day
busiest_day = max(day_count, key=day_count.get)
print("Busiest Day:", busiest_day)
print("Messages:", day_count[busiest_day])
hour_count = {}
# Initialize all 24 hours
for i in range(24):
    hour_count[i] = 0
for msg in messages:
    time = msg["timestamp"].split(",")[1].strip()
    hour = int(time.split(":")[0])
    hour_count[hour] += 1
# Find busiest hour
busiest_hour = max(hour_count, key=hour_count.get)
print("Busiest Hour:", busiest_hour)
print("Messages:", hour_count[busiest_hour])
print("=" * 60)
print("MOST ACTIVE DAY & HOUR")
print("=" * 60)
print(f"Busiest Day  : {busiest_day} ({day_count[busiest_day]} messages)")
print(f"Busiest Hour : {busiest_hour:02d}:00 - {busiest_hour+1:02d}:00 ({hour_count[busiest_hour]} messages)")
#Feature 4
import numpy as np
# AI-assisted
# Get participant names and sort them alphabetically
participants = sorted(list(participants))
# Dictionary to map each participant to a row index
person_index = {}
for i, person in enumerate(participants):
    person_index[person] = i
# Create a 6 x 24 matrix
activity_matrix = np.zeros((len(participants), 24), dtype=int)
# Fill the matrix
for msg in messages:
    sender = msg["sender"]
    # Extract hour from timestamp
    time = msg["timestamp"].split(",")[1].strip()
    hour = int(time.split(":")[0])
    row = person_index[sender]
    activity_matrix[row][hour] += 1
    print("Rows :", participants)
print()
print(activity_matrix)
print("=" * 90)
print("ACTIVITY MATRIX")
print("=" * 90)
header = "Name".ljust(12)
for h in range(24):
    header += f"{h:02d} "
print(header)
for i, person in enumerate(participants):
    row = person.ljust(12)
    for value in activity_matrix[i]:
        row += f"{value:02d} "
    print(row)
    print("=" * 90)
print("ACTIVITY HEATMAP")
print("=" * 90)
print("Hour :", end=" ")
for h in range(24):
    print(f"{h:02d}", end=" ")
print()
for i, person in enumerate(participants):
    print(f"{person:<10}", end=" ")
    maximum = max(activity_matrix[i])
    if maximum == 0:
        maximum = 1
    for value in activity_matrix[i]:
        ratio = value / maximum
        if ratio == 0:
            symbol = "."
        elif ratio <= 0.25:
            symbol = "░"
        elif ratio <= 0.50:
            symbol = "▒"
        else:
            symbol = "█"
        print(symbol, end="  ")
    print()
#Feature 5 # AI-assisted
stop_words = {
    "i", "is", "the", "a", "an", "and", "or", "to", "of", "in",
    "on", "for", "at", "it", "this", "that", "are", "am", "was",
    "were", "be", "been", "have", "has", "had", "do", "does",
    "did", "you", "your", "we", "our", "they", "their", "he",
    "she", "his", "her", "my", "me", "us", "with", "so", "if",
    "but", "as", "not", "just", "all", "can", "will", "its"
}
import string
word_count = {}
for msg in messages:
    # Skip media and deleted messages
    if msg["media"] or msg["deleted"]:
        continue
    text = msg["message"].lower()
    words = text.split()
    for word in words:
        # Remove punctuation
        word = word.strip(string.punctuation)
        if word == "":
            continue
        if word in stop_words:
            continue
        if word not in word_count:
            word_count[word] = 0
        word_count[word] += 1
        top_words = sorted(
    word_count.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]
print("=" * 60)
print("TOP 10 MOST USED WORDS")
print("=" * 60)
max_count = top_words[0][1]
for word, count in top_words:
    bar_length = int((count / max_count) * 25)
    bar = "█" * bar_length
    print(f"{word:<15} {bar} {count}")
#Feature 6
from datetime import datetime
for msg in messages:
    msg["datetime"] = datetime.strptime(
        msg["timestamp"], # AI-assisted
        "%d/%m/%y, %H:%M"
    )
    response_times = {}
for person in participants:
    response_times[person] = []
for i in range(1, len(messages)):
    previous = messages[i - 1]
    current = messages[i]
    # Ignore if the same person sent consecutive messages
    if previous["sender"] == current["sender"]:
        continue
    gap = (current["datetime"] - previous["datetime"]).total_seconds() / 60
    response_times[current["sender"]].append(gap)
    average_response = {}
for person in participants:
    times = response_times[person]
    if len(times) == 0:
        average_response[person] = 0
    else:
        average_response[person] = sum(times) / len(times)
        print("=" * 60)
print("AVERAGE RESPONSE TIME") # AI Assisted
print("=" * 60)
sorted_response = sorted(
    average_response.items(),
    key=lambda x: x[1]
)
for person, avg in sorted_response:
    print(f"{person:<10} {avg:.2f} minutes")
    active_days = {}
for person in participants:
    active_days[person] = set()
for msg in messages:
    active_days[msg["sender"]].add(msg["datetime"].date())
    start_date = messages[0]["datetime"].date()
end_date = messages[-1]["datetime"].date()
from datetime import timedelta
silent_streak = {}
for person in participants:
    longest = 0
    current = 0
    day = start_date
    while day <= end_date:
        if day in active_days[person]:
            current = 0
        else:
            current += 1
            if current > longest:
                longest = current
        day += timedelta(days=1)
    silent_streak[person] = longest
    print("=" * 60)
print("LONGEST SILENT STREAK")
print("=" * 60) #AI assisted
sorted_streak = sorted(
    silent_streak.items(),
    key=lambda x: x[1],
    reverse=True
)
for person, streak in sorted_streak:
    print(f"{person:<10} {streak} days")
#Feature 7
night_percentage = {}
for i, person in enumerate(participants):
    total_messages = activity_matrix[i].sum()
    night_messages = ( # AI assisted
        activity_matrix[i][23] +
        activity_matrix[i][0] +
        activity_matrix[i][1] +
        activity_matrix[i][2] +
        activity_matrix[i][3] +
        activity_matrix[i][4]
    )
    if total_messages == 0:
        night_percentage[person] = 0
    else:
        night_percentage[person] = (night_messages / total_messages) * 100
archetypes = {}
# Highest message sender
most_messages = max(message_count, key=message_count.get)
# Fastest responder
fastest = min(average_response, key=average_response.get)
# Longest silent streak
ghost = max(silent_streak, key=silent_streak.get)
for person in participants:
    if person == most_messages:
        archetypes[person] = "Spammer"
    elif night_percentage[person] >= 75:
        archetypes[person] = " Night Owl"
    elif person == fastest:
        archetypes[person] = "Fast Responder"
    elif person == ghost:
        archetypes[person] = " Ghost"
    elif media_count > 0:
        archetypes[person] = " Friendly"
    else:
        archetypes[person] = "Balanced"
        print("=" * 70)
print("PERSONALITY ARCHETYPES")
print("=" * 70)
for person in participants:
    print(f"{person:<10} {archetypes[person]}")

#Feature 8 #AI Assisted
print("=" * 80)
print("                    GROUPDNA - WHATSAPP CHAT ANALYZER")
print("=" * 80)
print("\nGROUP OVERVIEW")
print("-" * 80)
print(f"Total Participants : {len(participants)}")
print(f"Total Messages     : {len(messages)}")
print(f"Media Messages     : {media_count}")
print(f"Deleted Messages   : {deleted_count}")
print(f"System Messages    : {system_count}")
print("\n ACTIVITY SUMMARY")
print("-" * 80)
print(f"Busiest Day        : {busiest_day}")
print(f"Messages           : {day_count[busiest_day]}")
print(f"Busiest Hour       : {busiest_hour:02d}:00 - {(busiest_hour+1)%24:02d}:00")
print(f"Messages           : {hour_count[busiest_hour]}")
print("\n TOP MESSAGE CONTRIBUTORS")
print("-" * 80)
for person, count in sorted_people:
    percent = (count / len(messages)) * 100
    print(f"{person:<20} {count:>5} messages ({percent:.2f}%)")
print("\n⚡ AVERAGE RESPONSE TIME")
print("-" * 80)
for person, avg in sorted_response:
    print(f"{person:<20} {avg:.2f} minutes")
print("\n LONGEST SILENT STREAK")
print("-" * 80)
for person, streak in sorted_streak:
    print(f"{person:<20} {streak} days")
print("\n TOP 10 WORDS")
print("-" * 80)
for word, count in top_words:
    print(f"{word:<20} {count}")
print("\n PERSONALITY ARCHETYPES")
print("-" * 80)
for person in participants:
    print(f"{person:<20} {archetypes[person]}")
print("\n" + "=" * 80)
print("                 END OF GROUPDNA ANALYTICS REPORT")
print("=" * 80)





```
