```python
# Feature 1
# AI-assisted: Created a working copy of the original dataset
df_clean = df.copy()
print("Working copy created successfully.")
# Parse all mixed date formats
def parse_date(date_value):
    try:
        return pd.to_datetime(date_value, format="mixed", dayfirst=True)
    except:
        return pd.NaT
df_clean["Date"] = df_clean["Date"].apply(parse_date)
invalid_dates = df_clean["Date"].isna().sum()
print("Date parsing completed.")
print("Unparseable dates:", invalid_dates)
# Clean Amount column
df_clean["Amount"] = (
    df_clean["Amount"]
        .astype(str)
        .str.replace("₹", "", regex=False)
        .str.replace("Rs.", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
)
df_clean["Amount"] = pd.to_numeric(
    df_clean["Amount"],
    errors="coerce"
)
invalid_amounts = df_clean["Amount"].isna().sum()
print("Amount parsing completed.")
print("Unparseable amounts:", invalid_amounts)
# Standardize transaction type
df_clean["Type"] = (
    df_clean["Type"]
        .astype(str)
        .str.strip()
        .str.lower()
)
type_mapping = {
    "dr": "debit",
    "debit": "debit",
    "cr": "credit",
    "credit": "credit"
}
df_clean["Type"] = df_clean["Type"].map(type_mapping)
print("Transaction types standardized.\n")
print(df_clean["Type"].value_counts())
# Treat blank Mode values as missing
df_clean["Mode"] = (
    df_clean["Mode"]
        .replace("", np.nan)
)
print("Missing Mode values:", df_clean["Mode"].isna().sum())
# Remove duplicate transactions
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates()
rows_after = len(df_clean)
duplicates_removed = rows_before - rows_after
print("Duplicate rows removed:", duplicates_removed)
#AI Assisted Remove rows with invalid dates or amounts
rows_before = len(df_clean)
df_clean = df_clean.dropna(subset=["Date", "Amount"])
rows_after = len(df_clean)
removed_rows = rows_before - rows_after
print("Rows removed due to invalid Date/Amount:", removed_rows)
print("=" * 70)
print("FEATURE 1 VALIDATION")
print("=" * 70)
print(f"Original Transactions : {len(df)}")
print(f"Clean Transactions    : {len(df_clean)}")
print(f"Duplicates Removed    : {duplicates_removed}")
print(f"Date Parsing Errors   : {invalid_dates}")
print(f"Amount Parsing Errors : {invalid_amounts}")
print("\nTransaction Types")
print(df_clean["Type"].value_counts())
print("\nMissing Values")
print(df_clean.isnull().sum())
print("\nFeature 1 completed ")
```

```python
# Feature 2
print("=" * 70)
print("FEATURE 2 - VENDOR EXTRACTION")
print("=" * 70)
print("Unique Transaction Descriptions :", df_clean["Description"].nunique())
print("\nSample Descriptions\n")
for desc in sorted(df_clean["Description"].unique())[:100]:
    print(desc)
vendor_keywords = {
    # Food Delivery
    "Swiggy": ["SWIGGY", "BUNDL"],
    "Zomato": ["ZOMATO"],
    # Quick Commerce
    "Blinkit": ["BLINKIT", "GROFERS"],
    "Zepto": ["ZEPTO"],
    "Instamart": ["INSTAMART"],
    # Grocery
    "BigBasket": ["BIGBASKET", "KIRANAKART"],
    "DMart": ["DMART", "AVENUE SUPERMARTS", "INNOVATIVE RETAIL"],
    # Shopping
    "Amazon": ["AMAZON", "AMZN"],
    "Flipkart": ["FLIPKART", "FKART"],
    "Myntra": ["MYNTRA"],
    "Nykaa": ["NYKAA", "FSN"],
    "Ajio": ["AJIO"],
    # Entertainment
    "Netflix": ["NETFLIX"],
    "Spotify": ["SPOTIFY"],
    "Hotstar": ["HOTSTAR", "DISNEY", "STAR INDIA"],
    "BookMyShow": ["BOOKMYSHOW", "BIGTREE", "BMS"],
    # Investments
    "Groww": ["GROWW", "NEXTBILLION"],
    "Zerodha": ["ZERODHA"],
    # Transport
    "Uber": ["UBER"],
    "Ola": ["OLA", "ANI TECHNOLOGIES"],
    "Rapido": ["RAPIDO", "ROPPEN"],
    "BMTC": ["BMTC"],
    # Fuel
    "Indian Oil": ["INDIAN OIL", "IOC"],
    "HP Petrol": ["HP PETROL", "HPCL"],
    "BPCL": ["BPCL"],
    # Utilities
    "Airtel": ["AIRTEL", "BHARTI"],
    "Jio": ["JIO", "JIOFIBER"],
    "BESCOM": ["BESCOM", "BANGALORE ELEC"],
    "BWSSB": ["BWSSB"],
    "Vi": ["VI POSTPAID", "VODAFONE IDEA", "VI-RECHARGE"],
    # Cafes
    "Starbucks": ["STARBUCKS"],
    "Cafe Coffee Day": ["COFFEE DAY", "CCD"],
    "Third Wave Coffee": ["THIRD WAVE", "THIRDWAVE", "TWC INDIA"],
    # Restaurants
    "Empire Restaurant": ["EMPIRE"],
    "Meghana Foods": ["MEGHANA"],
    "Truffles": ["TRUFFLES"],
    "Restaurant": ["RESTAURANT"],
    "Dineout": ["DINEOUT"],
    # Income
    "Salary": ["SALARY", "TECHCRUSH"],
    # Rent
    "Rent": ["LANDLORD", "RENT"],
    # Cash
    "Cash Withdrawal": ["ATM-WDL", "ATM", "WDL"]
}
def extract_vendor(description):
    text = str(description).upper().strip()
    for vendor, keywords in vendor_keywords.items():
        for keyword in keywords:
            if keyword in text:
                return vendor
    if text.startswith("UPI-"):
        if "IOC" in text:
            return "Indian Oil"
        if "THIRDWAVE" in text:
            return "Third Wave Coffee"
        if "RESTAURANT" in text:
            return "Restaurant"
        if "VI" in text:
            return "Vi"
        return "Personal Transfer"
    return "Uncategorised"
if "vendor_clean" in df_clean.columns:
    df_clean.drop(columns=["vendor_clean"], inplace=True)
df_clean["vendor_clean"] = df_clean["Description"].astype(str).apply(extract_vendor)
print("Vendor extraction completed.")
print("="*70)
print("FEATURE 2 VALIDATION")
print("="*70)
print("Unique Canonical Vendors :", df_clean["vendor_clean"].nunique())
print("\nTop 20 Vendors\n")
print(df_clean["vendor_clean"].value_counts().head(20))
uncategorised = df_clean[df_clean["vendor_clean"]=="Uncategorised"]
print("="*70)
print("UNCATEGORISED AUDIT")
print("="*70)
print("Total Uncategorized :", len(uncategorised))
if len(uncategorised):
    print("\nDescriptions\n")
    for d in sorted(uncategorised["Description"].unique()):
        print(d)
else:
    print("No Uncategorized Vendors")

```

```python

# FEATURE 3
#AI Assisted
category_mapping = {
    "Swiggy": "Food",
    "Zomato": "Food",
    "Restaurant": "Food",
    "Empire Restaurant": "Food",
    "Meghana Foods": "Food",
    "Truffles": "Food",
    "Dineout": "Food",
    "Blinkit": "Groceries",
    "Zepto": "Groceries",
    "Instamart": "Groceries",
    "BigBasket": "Groceries",
    "DMart": "Groceries",
    "Amazon": "Shopping",
    "Flipkart": "Shopping",
    "Myntra": "Shopping",
    "Nykaa": "Shopping",
    "Ajio": "Shopping",
    "Netflix": "Entertainment",
    "Spotify": "Entertainment",
    "Hotstar": "Entertainment",
    "BookMyShow": "Entertainment",
    "Uber": "Transport",
    "Ola": "Transport",
    "Rapido": "Transport",
    "BMTC": "Transport",
    "Indian Oil": "Fuel",
    "HP Petrol": "Fuel",
    "BPCL": "Fuel",
    "Airtel": "Bills",
    "Jio": "Bills",
    "Vi": "Bills",
    "BESCOM": "Bills",
    "BWSSB": "Bills",
    "Starbucks": "Cafe",
    "Cafe Coffee Day": "Cafe",
    "Third Wave Coffee": "Cafe",
    "Groww": "Investment",
    "Zerodha": "Investment",
    "Salary": "Income",
    "Rent": "Housing",
    "Cash Withdrawal": "Cash",
    "Personal Transfer": "Transfer"
}
def assign_category(vendor):
    return category_mapping.get(vendor, "Others")
    # Remove old category column if it exists
if "category" in df_clean.columns:
    df_clean.drop(columns=["category"], inplace=True)
# Create new category column
df_clean["category"] = df_clean["vendor_clean"].apply(assign_category)
print("Category column created .")
print(df_clean.columns.tolist())
assert "category" in df_clean.columns, "Category column not created!"
print(df_clean[["vendor_clean", "category"]].head())
print("=" * 70)
print("FEATURE 3 VALIDATION")
print("=" * 70)
print("Unique Categories :", df_clean["category"].nunique())
print("\nCategory Distribution\n")
print(df_clean["category"].value_counts())
print("\nSample Records\n")
display(
    df_clean[
        ["Description", "vendor_clean", "category", "Amount"]
    ].head(10)
)


```

```python
# Feature 4
df_clean["Month"] = df_clean["Date"].dt.to_period("M").astype(str)
print("Month column created.")
# Separate income and expenses
monthly_income = (
    df_clean[df_clean["Type"] == "credit"]
    .groupby("Month")["Amount"]
    .sum()
)
monthly_expense = (
    df_clean[df_clean["Type"] == "debit"]
    .groupby("Month")["Amount"]
    .sum()
)
monthly_summary = (
    pd.DataFrame({
        "Income": monthly_income,
        "Expense": monthly_expense
    })
    .fillna(0)
)
monthly_summary["Savings"] = (
    monthly_summary["Income"] - monthly_summary["Expense"]
)
monthly_summary = monthly_summary.round(2)
print("Monthly summary created.")
category_monthly = (
    df_clean[df_clean["Type"] == "debit"]
    .pivot_table(
        values="Amount",
        index="Month",
        columns="category",
        aggfunc="sum",
        fill_value=0
    )
    .round(2)
)
print("Category-wise monthly expense table created.")
top_vendors = (
    df_clean[df_clean["Type"] == "debit"]
    .groupby("vendor_clean")["Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)
print("Top vendors calculated.")
print("=" * 70)
print("FEATURE 4 - MONTHLY ANALYTICS")
print("=" * 70)
print(f"Total Transactions : {len(df_clean)}")
print(f"Total Debit Amount : ₹{df_clean[df_clean['Type']=='debit']['Amount'].sum():,.2f}")
print(f"Total Credit Amount : ₹{df_clean[df_clean['Type']=='credit']['Amount'].sum():,.2f}")
print(f"Average Debit : ₹{df_clean[df_clean['Type']=='debit']['Amount'].mean():,.2f}")
print(f"Average Credit : ₹{df_clean[df_clean['Type']=='credit']['Amount'].mean():,.2f}")
print("=" * 70)
print("MONTHLY INCOME VS EXPENSE")
print("=" * 70)
display(monthly_summary)
print("=" * 70)
print("CATEGORY-WISE MONTHLY EXPENSE")
print("=" * 70)
display(category_monthly)
print("=" * 70)
print("TOP 10 VENDORS BY SPENDING")
print("=" * 70)
display(top_vendors)
```

```python
# FEATURE 5
monthly_trend = (
    df_clean[df_clean["Type"] == "debit"]
    .groupby("Month")["Amount"]
    .agg(
        Total_Spending="sum",
        Transactions="count",
        Average_Spending="mean"
    )
    .round(2)
)
print("Monthly trend created successfully.")
monthly_category = (
    df_clean[df_clean["Type"] == "debit"]
    .pivot_table(
        index="Month",
        columns="category",
        values="Amount",
        aggfunc="sum",
        fill_value=0
    )
    .round(2)
)
print("Monthly category trend created.")
monthly_trend["Previous_Month"] = monthly_trend["Total_Spending"].shift(1)
# AI Assisted
monthly_trend["Change"] = (
    monthly_trend["Total_Spending"] -
    monthly_trend["Previous_Month"]
).round(2)
monthly_trend["Trend"] = monthly_trend["Change"].apply(
    lambda x: "Increase" if x > 0 else ("Decrease" if x < 0 else "No Change")
)
monthly_trend = monthly_trend.fillna("-")
print("=" * 70)
print("FEATURE 5 - MONTHLY TRENDS")
print("=" * 70)
display(monthly_trend)
print("\nCATEGORY-WISE MONTHLY TREND\n")
display(monthly_category)
print("=" * 70)
print("FEATURE 5 VALIDATION")
print("=" * 70)
print("Months Analysed :", monthly_trend.shape[0])
print("Trend Direction\n")
print(monthly_trend[["Total_Spending", "Trend"]])
```

```python
# FEATURE 6
#AI Assisted
def get_time_bucket(time_str):
    hour = int(str(time_str).split(":")[0])
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"
df_clean["Time_Bucket"] = df_clean["Time"].apply(get_time_bucket)
print("Time buckets created.")
time_summary = (
    df_clean
    .groupby("Time_Bucket")
    .size()
    .rename("Transactions")
    .to_frame()
)
print(time_summary)
time_analysis = (
    df_clean
    .groupby(["Time_Bucket","Type"])["Amount"]
    .agg(["count","sum","mean"])
    .round(2)
)
display(time_analysis)
busiest = (
    df_clean["Time_Bucket"]
    .value_counts()
)
print("="*60)
print("FEATURE 5 ANALYSIS")
print("="*60)
print("Busiest Time Bucket :", busiest.idxmax())
print("Transactions        :", busiest.max())
category_time = (
    pd.crosstab(
        df_clean["Time_Bucket"],
        df_clean["category"]
    )
)
display(category_time)
print("="*60)
print("VALIDATION")
print("="*60)
print("Total Transactions :", len(df_clean))
print("Bucket Total       :", time_summary["Transactions"].sum())
```

```python
# FEATURE 7

debit_df = df_clean[df_clean["Type"] == "debit"].copy()
mean_amount = debit_df["Amount"].mean()
std_amount = debit_df["Amount"].std()
debit_df["Z_Score"] = (
    (debit_df["Amount"] - mean_amount)
    / std_amount
)
print("Z-score calculated.")
anomalies = debit_df[
    debit_df["Z_Score"].abs() >= 3
].copy()
print("Anomaly detection completed.")
anomalies = anomalies.sort_values(
    by="Amount",
    ascending=False
)
print("=" * 70)
print("FEATURE 7 - TOP ANOMALIES")
print("=" * 70)
display(
    anomalies[
        [
            "Date",
            "Time",
            "Description",
            "vendor_clean",
            "category",
            "Amount",
            "Z_Score"
        ]
    ].head(20)
)
#AI Assisted
print("=" * 70)
print("ANOMALY SUMMARY")
print("=" * 70)
print("Mean Amount        : ₹{:.2f}".format(mean_amount))
print("Standard Deviation : ₹{:.2f}".format(std_amount))
print()
print("Total Debit Transactions :", len(debit_df))
print("Total Anomalies          :", len(anomalies))
vendor_anomaly = (
    anomalies
    .groupby("vendor_clean")
    .agg(
        Transactions=("Amount", "count"),
        Total_Amount=("Amount", "sum")
    )
    .sort_values(
        by="Total_Amount",
        ascending=False
    )
)
print("=" * 70)
print("ANOMALOUS VENDORS")
print("=" * 70)
display(vendor_anomaly)
print("=" * 70)
print("FEATURE 7 VALIDATION")
print("=" * 70)
print("Formula Used")
print("Z = (Amount - Mean) / Standard Deviation")
print()
print("Maximum Z-Score :", round(debit_df["Z_Score"].max(), 2))
print("Minimum Z-Score :", round(debit_df["Z_Score"].min(), 2))
print()
print("Top 10 Largest Transactions")
display(
    debit_df.nlargest(
        10,
        "Amount"
    )[
        [
            "Description",
            "vendor_clean",
            "Amount",
            "Z_Score"
        ]
    ]
)
```

```python
# FEATURE 8
if "Hour" not in df_clean.columns:
    df_clean["Hour"] = (
        df_clean["Time"]
        .astype(str)
        .str.split(":")
        .str[0]
        .astype(int)
    )
profile = {}
profile["food_spend"] = df_clean[df_clean["category"]=="Food"]["Amount"].sum()
profile["shopping_spend"] = df_clean[df_clean["category"]=="Shopping"]["Amount"].sum()
profile["investment_spend"] = df_clean[df_clean["category"]=="Investment"]["Amount"].sum()
profile["transport_spend"] = df_clean[df_clean["category"]=="Transport"]["Amount"].sum()
profile["cafe_spend"] = df_clean[df_clean["category"]=="Cafe"]["Amount"].sum()
profile["bill_spend"] = df_clean[df_clean["category"]=="Bills"]["Amount"].sum()
profile["late_night_orders"] = len(
    df_clean[
        (df_clean["Hour"]>=21) &
        (df_clean["vendor_clean"].isin(["Swiggy","Zomato"]))
    ]
)
profile["amazon_orders"] = len(df_clean[df_clean["vendor_clean"]=="Amazon"])
profile["swiggy_orders"] = len(df_clean[df_clean["vendor_clean"]=="Swiggy"])
profile["zomato_orders"] = len(df_clean[df_clean["vendor_clean"]=="Zomato"])
print(profile)
def foodie(profile):
    return profile["food_spend"] > 70000
def investor(profile):
    return profile["investment_spend"] > 100000
def shopaholic(profile):
    return profile["shopping_spend"] > 150000
def commuter(profile):
    return profile["transport_spend"] > 40000
def cafe_lover(profile):
    return profile["cafe_spend"] > 20000
def night_owl(profile):
    return profile["late_night_orders"] >= 10
archetypes = []
if foodie(profile):
    archetypes.append("Foodie")
if investor(profile):
    archetypes.append("Investor")
if shopaholic(profile):
    archetypes.append("Shopaholic")
if commuter(profile):
    archetypes.append("Daily Commuter")
if cafe_lover(profile):
    archetypes.append("Cafe Lover")
if night_owl(profile):
    archetypes.append("Night Owl")
    print("="*70)
print("FEATURE 8 - ARCHETYPE DETECTION")
print("="*70)
print("Detected Archetypes\n")
for i, archetype in enumerate(archetypes,1):
    print(f"{i}. {archetype}")
    print("\n"+"="*70)
print("SUPPORTING METRICS")
print("="*70)
print(f"Food Spend           : ₹{profile['food_spend']:,.2f}")
print(f"Shopping Spend       : ₹{profile['shopping_spend']:,.2f}")
print(f"Investment Spend     : ₹{profile['investment_spend']:,.2f}")
print(f"Transport Spend      : ₹{profile['transport_spend']:,.2f}")
print(f"Cafe Spend           : ₹{profile['cafe_spend']:,.2f}")
print(f"Bills                : ₹{profile['bill_spend']:,.2f}")
print()
print(f"Amazon Orders        : {profile['amazon_orders']}")
print(f"Swiggy Orders        : {profile['swiggy_orders']}")
print(f"Zomato Orders        : {profile['zomato_orders']}")
print(f"Late Night Orders    : {profile['late_night_orders']}")
print("="*70)
print("FEATURE 8 VALIDATION")
print("="*70)
print("Functions Used")
print("foodie()")
print("investor()")
print("shopaholic()")
print("commuter()")
print(" cafe_lover()")
print("night_owl()")
print()
print("Total Archetypes Detected :", len(archetypes))
```
